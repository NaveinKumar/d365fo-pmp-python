# Firm Fixed Price (FFP)

## D365FO Project Accounting + PMP + Python

A Firm Fixed Price (FFP) contract is an agreement where the buyer
agrees to pay a fixed price for a specified product or service.

The seller bears the risk of cost overruns or unexpected expenses.

In this notebook, we will use Python to visualize this risk.


```python
contract_price = 100_000
estimated_cost = 80_000

estimated_profit = contract_price - estimated_cost

print("Contract price:", contract_price)
print("Estimated cost:", estimated_cost)
print("Estimated profit:", estimated_profit)
```

## Actual Cost and Actual Profit

The estimated profit is based on our original cost estimate.

In a real project, the actual cost can be different.

Because the FFP contract price is fixed, any change in actual
cost directly affects the seller's profit.


```python
actual_cost = 90_000

actual_profit = contract_price - actual_cost

print("Contract price:", contract_price)
print("Actual cost:", actual_cost)
print("Actual profit:", actual_profit)
```


## Visualizing FFP Risk

The contract price remains fixed at $100,000.

As the seller's actual cost increases, the seller's profit decreases.

Let's visualize this relationship.


```python
import matplotlib.pyplot as plt

actual_costs = [60_000, 70_000, 80_000, 90_000, 100_000, 110_000, 120_000]

profits = []

for cost in actual_costs:
    profit = contract_price - cost
    profits.append(profit)

plt.plot(actual_costs, profits)
plt.xlabel("Actual Cost")
plt.ylabel("Seller Profit")
plt.title("FFP: Actual Cost vs Seller Profit")

plt.show()
```

```python
plt.plot(actual_costs, profits)

plt.axhline(0)

plt.xlabel("Actual Cost")
plt.ylabel("Seller Profit")
plt.title("FFP: Actual Cost vs Seller Profit")

plt.show()
```

## Think About It

The contract price is fixed at $100,000.

At what actual cost does the seller:

- make a profit?
- break even?
- begin losing money?

What happens if the actual cost reaches $120,000?

## Let Python Calculate Every Scenario

Instead of calculating each cost scenario manually, we can ask Python
to calculate the profit for every possible actual cost in our list.

This is what a `for` loop allows us to do.

```python
for cost in actual_costs:
    profit = contract_price - cost
    print("Cost:", cost, "Profit:", profit)
```

## What Python Just Did

Python repeated the same calculation for every cost in our list:

    Profit = Contract Price - Actual Cost

This lets us examine many project scenarios quickly.

For an FFP contract:

- Actual cost below $100,000 → profit
- Actual cost of $100,000 → break-even
- Actual cost above $100,000 → loss

The contract price did not change.

The seller's financial result changed because the actual cost changed.
